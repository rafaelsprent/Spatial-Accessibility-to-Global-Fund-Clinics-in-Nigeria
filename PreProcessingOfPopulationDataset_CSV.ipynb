# PreProcessing of Population Density Dataset

```python
import geopandas as gpd
import pandas as pd
```

```python
# Read in the excel to a dataframe
nigeria_population_density_pd = pd.read_csv("Datasets/nga_general_2020_csv/nga_general_2020.csv")
nigeria_population_density_pd.head()
```

```python
# Convert to geopandas dataframe
nigeria_population_density_gpd = gpd.GeoDataFrame(nigeria_population_density_pd, 
    geometry = gpd.points_from_xy(nigeria_population_density_pd['longitude'], nigeria_population_density_pd['latitude']), 
    crs = 'EPSG:4326')
```

```python
# Validate that the clinics are admin boundaries are correct visually
nigeria_population_density_gpd_extract = nigeria_population_density_gpd.sample(n=1000, random_state=1) 
folium_map = nigeria_population_density_gpd_extract.explore()
```
