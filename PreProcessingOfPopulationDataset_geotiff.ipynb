# PreProcessing of Population Density Dataset

```python
# Import raster and add to folium map
import rasterio
from rasterio.plot import reshape_as_image
import folium
from folium.raster_layers import ImageOverlay
import numpy as np
```

```python
# Path to raster file (update with actual path if needed)
raster_path = 'Datasets\\nga_general_2020_geotiff\\nga_general_2020.tif'

# Read raster data
with rasterio.open(raster_path) as src:
    raster = src.read(1)
    bounds = src.bounds
    transform = src.transform

# Get image extent for folium
image_extent = [
    [bounds.bottom, bounds.left],
    [bounds.top, bounds.right]
]
```

```python
# get all the unique values in the raster
unique_values = np.unique(raster)
print("Unique values in the raster:", unique_values)
# Change anything below 0 or 0 to np.nan
raster = np.where(raster <= 0, np.nan, raster)
```

```python

# Create folium map centered on Nigeria
m = folium.Map(location=[9.082, 8.6753], zoom_start=6)
```

```python
# Add raster as image overlay
ImageOverlay(
    image=raster,
    bounds=image_extent,
    opacity=0.6,
    interactive=True,
    cross_origin=False,
    zindex=1
).add_to(m)

# Display map
m
```
