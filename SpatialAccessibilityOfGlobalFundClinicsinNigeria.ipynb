# Spatial Accessibility of Global Fund Clincs in Nigeria

## Required Packages

```python
import geopandas as gpd
import pandas as pd
import requests
```

## Datasets

Population Density Dataset: https://data.humdata.org/dataset/highresolutionpopulationdensitymaps-nga

Global Fund Clinic Data : Global Fund 

Nigeria Administrative Boundaries: https://data.humdata.org/dataset/cod-ab-nga

## PreProcessing

### Admin Boundaries 

Extract geojson and convert to geopandas

```python
nigeria_admin_boundaries_gpd = gpd.read_file("Datasets/nga_admin_boundaries.geojson/nga_admin1.geojson")
nigeria_admin_boundaries_gpd.head()
```

### Global Fund Clincs

Extracting the clincs from the CSV and into a geopandas dataframe

```python
# Read in the excel to a dataframe
global_fund_clinics_pd = pd.read_excel("Datasets/LMD Input Data.xlsx",sheet_name='HL')
global_fund_clinics_pd.head()
```

```python
# Convert to geopandas dataframe
global_fund_clincs_gpd = gpd.GeoDataFrame(global_fund_clinics_pd, 
    geometry = gpd.points_from_xy(global_fund_clinics_pd['Destination Longitude'], global_fund_clinics_pd['Destination Latitude']), 
    crs = 'EPSG:4326')
```

```python
#V alidate that the clinics are admin boundaries are correct visually
folium_map = nigeria_admin_boundaries_gpd.explore()
global_fund_clincs_gpd.explore(m=folium_map, color='red', marker_kwds={'radius':1, 'fill':True})   
folium_map
```
